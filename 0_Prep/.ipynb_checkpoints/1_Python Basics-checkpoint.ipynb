## Conditional Statements

- if
- else

```python
a = 15

if a > 10:
    print("Hello")
```

```python
a = 15

if a > 100:
    print("Hello")
```

```python
a = 15

if a > 100:
    print("Hello")
else:
    print("Bye")
```

```python
pin = 1234

if pin == 1234:
    print("Your Balance: $100, Take your money!!")
else:
    print("Incorrect Pin")
```

```python
pin = int(input("Enter your pin: "))

if pin == 1234:
    print("Your Balance: $100, Take your money!!")
else:
    print("Incorrect Pin")
```

## While Loops

```python
print("Himanshu")
```

```python
i = 10

print(i > 10)
```

```python
i = 1

while i <= 10:
    print("Himanshu")
    i = i + 1
```

```python
i = 1

while i <= 10:
    print(i)
    i = i + 1
```
