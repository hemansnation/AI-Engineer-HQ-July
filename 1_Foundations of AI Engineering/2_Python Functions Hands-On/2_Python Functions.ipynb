```python
def add(x,y):
    print(x)
    print(y)
    print(x + y)

add(5,4)
```

```python
def add1(v):              # v is a local variable
    v = v + 1
    print(v)

add1(7)
print(v)
```

```python
x = 3                     # x is global variable

def add1(x):              # x is a local variable
    x = x + 1
    print(x)

add1(7)
print(x)
```

```python
def play(x, y):
    if x >= y:
        return x
    else:
        return y

print(play(10,5))
#         10
```

```python
print(play(15,20))
```
