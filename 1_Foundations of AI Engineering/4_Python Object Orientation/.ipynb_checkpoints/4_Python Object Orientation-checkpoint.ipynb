# Object Oriented Programming


7 properties

1 - class - updated version of structure
            collection of variables and methods(functions) 
            class is a blueprint

2 - object - run time or real time entity
             hashcode (java), id (python), virtual code (cpp)

3 - abstraction  - showing only important features without showing any background details

- encapsulation - wrapping up of data in a single unit

4 - inheritance - acquiring properties of one class into another
                - resuing the code

5 - polymorphism - same name multiple functionalities (method overloading and method overriding)

6 - dynamic memory allocation - run time memory allocation

7 - message passing - communication between objects

```python
class Test:

    def add(self):
        print(self)
        print("This is add method of class Text")

    def sub(self, a, b):          # a and b are local variables, x and y are class variables
        self.x = a                # t1.x = 10, t2.x = 11
        self.y = b                # t1.y = 20, t2.y = 22

    def show(self):
        print(self.x)
        print(self.y)
```

```python
t1 = Test()
t2 = Test()
```

```python
t1.add()
t2.add()
```

```python
t1.sub(10,20)
```

```python
t2.sub(11,22)
```

```python
t1.show()
```

```python
t2.show()
```

## Inheritance

1. single level
2. multilevel
3. hierarchical
4. multiple
5. hybrid



### single level



```python
A       - windows xp
|
B       - windows 7
```

```python
class A:                                    # parent class/base class/super class
    
    def add(self):
        print("this is add from class A")

class B(A):                                 # child class/derived class/ sub class

    def sub(self):
        print("this is sub from class B")
```

```python
x = A()
y = B()
```

```python
x.add()
y.add()
```

```python
y.sub()
```

```python
x.sub()
```

### inheritance with constuctor 

```python
class Windows7:
    def __init__(self):
        print("Init Windows7")

    def add(self):
        print("Add from windows7")

class Windows10(Windows7):
    def __init__(self):
        print("Init Windows10")

    def sub(self):
        print("Sub from windows10")
    
```

```python
x = Windows7()
y = Windows10()
```

```python
class Windows7:
    def __init__(self):
        print("Init Windows7")

    def add(self):
        print("Add from windows7")

class Windows10(Windows7):
    def __init__(self):
        Windows7.__init__(self)
        print("Init Windows10")

    def sub(self):
        print("Sub from windows10")
    
```

```python
x = Windows7()
y = Windows10()
```

### multilevel

```python
A
|
B
|
C
```

```python
class A:
    def add(self):
        print("Add A")

class B(A):
    def sub(self):
        print("Sub B")

class C(B):
    def show(self):
        print("Show C")

z = C()
z.show()
z.sub()
z.add()
```

## Hierarchical Inheritance

```python
    A
B       C
```

```python
class A:
    def add(self):
        print("Add A")

class B(A):
    def sub(self):
        print("Sub B")

class C(A):
    def show(self):
        print("Show C")

z = C()
z.show()
z.add()
z.sub()
```

## multiple inheritance

ambiguity error

C3 - algorithm to solve this error

MRO - Method Resolution Order

```python
A       B
    C
```

```python
class A:
    def add(self):
        print("Add A")

class B:
    def add(self):
        print("add B")

class C(A,B):
    def show(self):
        print("Show C")

z = C()
z.show()
```

```python
z.add()
```

```python
class A:
    def add(self):
        print("Add A")

class B:
    def add(self):
        print("add B")

class C(B,A):
    def show(self):
        print("Show C")

z = C()
z.add()
```

```python
print(C.__mro__)
```

## Method Overloading

```python
a = 10
a = 20

print(a)
```

```python
def play(a,b):
    print(a)
    print(b)

def play(x,y,z):
    print(x)
    print(y)
    print(z)
```

```python
play()
```

```python
play(10,20)
```

```python
play(10,20,30)
```

```python
print(type(play))
```

```python
print(id(play))
```
