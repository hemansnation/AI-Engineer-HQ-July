```python
# mean

sum of all the values / total number of values

- it best represents data
- it consider all the data points
- discrete data - obtained by counting
                    - number of students
                    - number of chairs in a classroom
                    - number of coins giving heads
- continuous data - obtained bt measuring
                    - height of students
                    - weights
                    - time it takes for a sesssion
                    - distance from class to home

- extremely sensitive to the presence of outliers


- normalization
- loss functions
- embeddings
- evaluation
```

```python
# median

- arrange the data points in ascending to descending order
- less sensitive to outliers
- 50% of the datais on either side of it

- decision trees
```

```python
# mode

- values that occurs most
- sentiment from a data of reviews - what is the frequency of positive words

- vote in an election
- mode is not good for continuous data
```

```python
# standard deviation


5.5 feet - mean height

depth of the river - 5 feet

sd - 1.2

range of the heights: 4.3 - 6.7

- distance from a data point to mean
- sum of all the distance from mean is standard deviation 

```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
forbes = pd.read_csv('datasets/forbes.csv')

forbes.head()
```

```python
forbes.isnull().sum()
```

```python
forbes.info()
```

```python
forbes.dropna(inplace=True)

forbes.isnull().sum()
```

```python
forbes.shape
```

```python
market_value_mean = forbes['Market Value'].mean()

market_value_mean
```

```python
market_value_median = forbes['Market Value'].median()

market_value_median
```

```python
forbes['Market Value'].min()
```

```python
forbes['Market Value'].max()
```

```python
forbes['Sector'].value_counts()
```

### box plot

- mean
- median
- 25th percentile
- 75th percentile
- maximum values
- outliers

```python
plt.figure(figsize=(12,8))

sns.boxplot(data=forbes, y = 'Market Value', showmeans=True)

plt.axhline(y=market_value_mean, color='g', linestyle='-')
plt.axhline(y=market_value_median, color='r', linestyle='-')

plt.title('Forbes market value data')

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.boxplot(data=forbes, y = 'Market Value', showmeans=True)

plt.ylim(0,80)

plt.axhline(y=market_value_mean, color='g', linestyle='-')
plt.axhline(y=market_value_median, color='r', linestyle='-')

plt.title('Forbes market value data')

plt.show()
```
